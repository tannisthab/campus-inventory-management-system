# Relational Database Schema Design and SQL Implementation

**Project:** Campus Inventory Management System

**Course:** 26DSMJA1CC01 – Relational Database Management System

**Technology Used:** MySQL Workbench, MySQL, Jupyter Notebook

## 1. Aim

To design and implement a relational database for a campus inventory management system using SQL. The project focuses on entity identification, relationship establishment, primary and foreign key constraints, data manipulation and multi-table join operations.

## 2. Objectives

- To identify entities and their attributes.
- To design an Entity-Relationship model.
- To implement relational tables using SQL DDL commands.
- To populate tables using SQL DML commands.
- To establish primary key and foreign key constraints.
- To perform data manipulation and multi-table JOIN operations.
- To generate an EER schema diagram using MySQL Workbench.

## 3. Problem Statement

A campus manages various resources such as computers, laboratory equipment, stationery, furniture and electronic devices. Maintaining inventory records manually can lead to difficulties in tracking available stock, supplier information, staff responsibilities and inventory transactions.

The Campus Inventory Management System is designed to organise this information using a relational database. It provides a structured way to store inventory details, monitor stock levels, maintain supplier records and retrieve transaction information through SQL queries.

## 5. Entity Identification

The database consists of six main entities.

| Entity | Description |
|---|---|
| Department | Stores campus department details |
| Supplier | Stores supplier information |
| Category | Classifies inventory items |
| Staff | Maintains staff details |
| Item | Stores inventory item information |
| Inventory_Transaction | Records stock-related transactions |

### 5.1 Entity Attributes

**Department:** department_id, department_name, location.

**Supplier:** supplier_id, supplier_name, contact, email.

**Category:** category_id, category_name.

**Staff:** staff_id, staff_name, department_id, designation.

**Item:** item_id, item_name, category_id, supplier_id, unit_price, quantity_in_stock, reorder_level.

**Inventory_Transaction:** transaction_id, item_id, staff_id, transaction_type, quantity, transaction_date.

## 6. Relationships Between Entities

The database uses the following relationships:

| Parent Entity | Child Entity | Relationship |
|---|---|---|
| Department | Staff | One-to-Many |
| Supplier | Item | One-to-Many |
| Category | Item | One-to-Many |
| Staff | Inventory_Transaction | One-to-Many |
| Item | Inventory_Transaction | One-to-Many |

### Explanation

- A department can have multiple staff members.
- A supplier can supply multiple inventory items.
- A category can contain multiple items.
- A staff member can record multiple inventory transactions.
- An inventory item can appear in multiple transactions.

## 7. Relational Database Schema

The relational schema is represented as follows:

1. **Department** (department_id, department_name, location)
   - Primary Key: department_id

2. **Supplier** (supplier_id, supplier_name, contact, email)
   - Primary Key: supplier_id

3. **Category** (category_id, category_name)
   - Primary Key: category_id

4. **Staff** (staff_id, staff_name, department_id, designation)
   - Primary Key: staff_id
   - Foreign Key: department_id references Department

5. **Item** (item_id, item_name, category_id, supplier_id, unit_price, quantity_in_stock, reorder_level)
   - Primary Key: item_id
   - Foreign Keys: category_id references Category; supplier_id references Supplier

6. **Inventory_Transaction** (transaction_id, item_id, staff_id, transaction_type, quantity, transaction_date)
   - Primary Key: transaction_id
   - Foreign Keys: item_id references Item; staff_id references Staff

Primary keys uniquely identify records, while foreign keys establish relationships between the tables and maintain referential integrity.

The EER diagram was generated using the Reverse Engineer feature in MySQL Workbench.

The diagram represents the six database tables, their attributes, primary keys, foreign keys and relationships.
